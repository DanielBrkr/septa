```python
from pathlib import Path

import pandas as pd
from autoviz import AutoViz_Class
from autoviz import data_cleaning_suggestions
from tqdm.auto import tqdm

tqdm.pandas()


%matplotlib inline
```

# Vital Aperiodic Exploration

```python
INTER_DATA_PATH = "../../intermediary_data/"
```

```python
filename_parq = "vital_aperiodic.parquet"
file_path_parq = Path(INTER_DATA_PATH) / Path(filename_parq)
```

```python
df = pd.read_parquet(path=file_path_parq, engine="pyarrow")

```

```python
df.info()
```

```python
AV = AutoViz_Class()
plot = False
if plot:
    
    AV.AutoViz(
        #filename=str(file_path_parq.resolve()),
        filename="",
        dfte=df,
        verbose=2,
        lowess=True,
        #chart_format="svg",
        chart_format='png',
        #depVar="diedinhospital",
       # max_cols_analyzed=52,
        max_rows_analyzed=1000000,
        save_plot_dir=r"vpa_plot_outputs"
    )
```

```python
df.head()
```

```python
def convert_to_timedelta(stay):
    stay["time_delta"] = pd.to_timedelta(stay["observationoffset"] - stay["observationoffset"].iloc[0], unit="min")

    return stay
```

```python
df = df.groupby("patientunitstayid").progress_apply(
    convert_to_timedelta).reset_index(drop=True)
```

```python
df.head()
```

```python
AV.AutoViz(
    #filename=str(file_path_parq.resolve()),
    filename="",
    dfte=df,
    verbose=2,
    lowess=True,
    #chart_format="svg",
    chart_format='png',
    #depVar="diedinhospital",
    # max_cols_analyzed=52,
    #max_rows_analyzed=1000000,
    save_plot_dir=r"vpa_plot_outputs"
)
```
