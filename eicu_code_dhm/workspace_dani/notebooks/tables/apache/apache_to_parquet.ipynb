```python
from pathlib import Path

import holoviews as hv
import hvplot.pandas  # noqa+
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from loguru import logger

from notebooks_dani.eicu_db_connection import DatabaseConnection
```

```python
INTER_DATA_PATH = Path("..") / Path("..") / Path("intermediary_data")
logger.info(f"Loading from {Path(INTER_DATA_PATH).resolve()}")
```

```python
apache_valid_results_path = INTER_DATA_PATH / Path("apache_aps_var_results_valid_mortality.csv")

df_apache_valid = pd.read_csv(filepath_or_buffer=apache_valid_results_path.resolve())
df_apache_valid.head(n=3)
```

```python
df_apache_valid.info()
```

```python
# int32
vars_int = ["intubated", "vent", "dialysis", "eyes", "motor", "verbal", "meds",
            "acutephysiologyscore", "apachescore", "actualventdays", "unabridgedactualventdays"]
# float32
vars_float = ["urine", "temperature", "respiratoryrate", "heartrate", "meanbp", "wbc", "sodium", "hematocrit", "creatinine", "albumin", "glucose", "bilirubin","ph", "pao2", "pco2", "fio2", "bun", "predictedicumortality", "predictediculos", "predictedhospitallos", "actualhospitallos", "unabridgedunitlos", "unabridgedhosplos", "predventdays"]

# cat
vars_cat = ["physicianspeciality", "physicianinterventioncategory", "apacheversion", "actualicumortality", "actualhospitalmortality", ]

# bool
vars_bool = ["preopmi", "preopcardiaccath", "ptcawithin24h" ]
```

```python
df_apache_valid[[*vars_int]] = df_apache_valid[[*vars_int]].apply(pd.to_numeric, downcast='integer')
df_apache_valid[[*vars_float]] = df_apache_valid[[*vars_float]].apply(pd.to_numeric, downcast='float')
df_apache_valid[[*vars_cat]] = df_apache_valid[[*vars_cat]].apply(pd.Categorical)
df_apache_valid[[*vars_bool]] = df_apache_valid[[*vars_bool]].astype("boolean")
```

```python
path = Path("..") /  Path("..") / Path("intermediary_data/apache") / Path("apache_aps_var_results_valid_mortality")
df_apache_valid.to_csv(path_or_buf=str(path.resolve()) + ".csv" , index=False)
```
