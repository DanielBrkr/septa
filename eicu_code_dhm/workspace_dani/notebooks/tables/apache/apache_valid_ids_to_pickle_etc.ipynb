```python
import connectorx as cx
import pandas as pd
from pathlib import Path
from notebooks_dani.eicu_db_connection import DatabaseConnection
from loguru import logger
import pickle
```

# This notebook gets the apache table with the valid entries and spits out the unique values of  "patientunistayids" for further use with other tables 

```python
connection = DatabaseConnection(config_path="../../db/config_container.ini")
engine = connection.engine
query_schema = connection.query_schema
conn_url = connection.database_url
```

```python
INTER_DATA_PATH = Path("..") / Path("..") / Path("intermediary_data")
INTER_DATA_PATH_SAVE_PATH = INTER_DATA_PATH / Path("valid_patient_unit_stay_ids")

logger.info(f"Loading from {INTER_DATA_PATH.resolve()}")
logger.info(f"Saving to {INTER_DATA_PATH_SAVE_PATH.resolve()}")
```

```python
apache_results_valid_filename = "apache_aps_var_results_valid_mortality.csv"
apache_results_valid_path = INTER_DATA_PATH / Path(apache_results_valid_filename)

df_apache_valid = pd.read_csv(filepath_or_buffer=apache_results_valid_path)
```

```python
df_apache_valid.info()
```

```python
df_apache_valid.head(n=10)
```

```python
patient_stays_valid: list = df_apache_valid["patientunitstayid"].unique().tolist()
sorted_patient_stays_valid = sorted(patient_stays_valid)
```

```python
sorted_patient_stays_valid[:10]
```

```python
valid_ids_filename_pickle = "valid_ids_from_apache.pkl"
valid_ids_path_pickle = INTER_DATA_PATH_SAVE_PATH / valid_ids_filename_pickle

with open(valid_ids_path_pickle, "wb") as f:
    pickle.dump(sorted_patient_stays_valid, f)
    
```

```python
ids_series: pd.Series = pd.Series(sorted_patient_stays_valid, name="patientunitstayid", dtype="Int64")
ids_df: pd.DataFrame = pd.DataFrame(sorted_patient_stays_valid, columns=["patientunitstayid"], dtype="Int64")
```

```python
valid_ids_path_csv = INTER_DATA_PATH_SAVE_PATH / Path("valid_ids_from_apache.csv")
valid_ids_path_parquet = INTER_DATA_PATH_SAVE_PATH / Path("valid_ids_from_apache.parquet")
```

```python
ids_series.to_csv(path_or_buf=valid_ids_path_csv)
```
