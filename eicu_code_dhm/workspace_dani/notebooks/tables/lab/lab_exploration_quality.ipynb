```python
import connectorx as cx
import pandas as pd
from pathlib import Path
from tqdm.auto import tqdm
from autoviz import AutoViz_Class
from autoviz import data_cleaning_suggestions

tqdm.pandas()

%matplotlib inline
```

```python
INTER_DATA_PATH = "../../intermediary_data/"
filename_csv = "lab.csv"
filename_parq = "lab.parquet"
file_path_csv = Path(INTER_DATA_PATH) / Path(filename_csv)
file_path_parq = Path(INTER_DATA_PATH) / Path(filename_parq)
```

```python
df = pd.read_parquet(path=file_path_parq, engine="pyarrow")
```

```python
df.info()
```

```python
df.shape[0]
```

```python
def convert_to_timedelta(stay):
    stay["time_delta"] = pd.to_timedelta(stay["labresultoffset"] - stay["labresultoffset"].iloc[0], unit="min")
    stay["time_delta_revised"] = pd.to_timedelta(stay["labresultrevisedoffset"] - stay["labresultrevisedoffset"].iloc[0], unit="min")

    return stay
```

```python
df = df.groupby("patientunitstayid").progress_apply(
    convert_to_timedelta).reset_index(drop=True)
```

```python
AV = AutoViz_Class()
plot = False
if plot:

    AV.AutoViz(
        #filename=str(file_path_parq.resolve()),
        filename="",
        dfte=df,
        verbose=2,
        lowess=True,
        #chart_format="svg",
        chart_format='png',
        #depVar="diedinhospital",
        # max_cols_analyzed=52,
        max_rows_analyzed=df.shape[0],
        save_plot_dir=r"lab_plot_outputs"
    )
```
