```python
df_heartrate = df["heartrate"].iloc[:1000].to_frame()
df_heartrate.columns = ["heartrate"]
```

```python
cA, cD = pywt.dwt(df_heartrate["heartrate"], 'db1')
```

```python
df_heartrate_reconstructed = pd.DataFrame(pywt.idwt(cA=cA, cD=None, wavelet="db1"), columns=['heartrate'])
```

```python
plot_raw = df_heartrate.hvplot(kind="line", label="raw",width=1080, height=500)
plot_reconstructed = df_heartrate_reconstructed.hvplot(kind="line", label="wavelet filtered")
```

```python
plot_raw * plot_reconstructed
```
