```python
import pandas as pd
import numpy as np
from loguru import logger
from workspace_dani.data_preprocessing.utils.data_handling import load_parquet_table, save_parquet_table
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns
import hvplot.pandas
import holoviews as hv

notebook_path = Path.cwd()
project_root = notebook_path.parents[2]

dataset_path: Path = project_root / Path("intermediary_data/dataset/sepsis-type-a-cohort/2024-07-02/dataset_ts.parquet")


plot_outputs = Path("plot_outputs")
```

```python
df_ds = load_parquet_table(parquet_path=dataset_path, verbose=False, backend="numpy_nullable")  #.head(n=5000)
df_ds.head()
```

### Correlation of mean values with being a sepsis patient

```python
df_means = df_ds.groupby("patientunitstayid").mean().reset_index()
```

```python
means_correlation_matrix = df_means.corr()
```

```python
plt.figure(figsize=(20, 20))
sns.heatmap(means_correlation_matrix, annot=False, cmap="icefire", cbar=True, square=True, linewidths=0.5)
plt.xticks(rotation=90, fontsize=8)
plt.yticks(rotation=0, fontsize=8)
# plt.xticks(ticks=[], labels=[])
# plt.yticks(ticks=[], labels=[])
plt.show()
```

```python
heatmap = means_correlation_matrix.hvplot.heatmap(
    title="Means vs. Sepsis Patient Correlation Matrix (Cohort A)",
    cmap="bkr",
    clim=(-1, 1),
    width=2000,
    height=2000,
).opts(xrotation=90, fontsize={
    "title": 20,  
    "labels": 5,  
    "xticks": 10,  
    "yticks": 10,  
})

hv.save(heatmap, plot_outputs / "Sepsis Cohort A Means Correlation Matrix.html")
```

```python
heatmap
```

```python
correlation_with_sepsis = df_means.corr()["sepsis_patient"].drop(columns="sepsis_patient")
correlation_with_sepsis.columns = "sepsis_corr"

plt.figure(figsize=(21, 10))

correlation_with_sepsis.sort_values(ascending=False).plot(kind="bar")

plt.title("Sepsis Patient Correlation")
plt.xlabel("Features")
plt.ylabel("Correlation Coefficient")
plt.xticks(rotation=90, fontsize=8)
plt.tight_layout()
plt.show()
```

```python
bar_plot_means = correlation_with_sepsis.sort_values(ascending=False).hvplot.bar(
    title="Means vs. Sepsis Patient Correlation (Cohort A)",
    xlabel="Features",
    ylabel="Correlation Coefficient",
    rot=90,
    width=1800,
    height=600
).opts(
    fontsize={
        "title": 20,
        "labels": 10,
        "xticks": 8,
        "yticks": 10
    }
)

hv.save(bar_plot_means, plot_outputs / "Sepsis Cohort A Means Correlation Bars.html")
```

### Correlation of std to sepsis

```python
df_stds = df_ds.drop(columns="sepsis_present").groupby(["patientunitstayid","sepsis_patient"]).std(numeric_only=True).reset_index()

stds_correlation_matrix = df_stds.corr()
```

```python
heatmap_stds = stds_correlation_matrix.hvplot.heatmap(
    title="Standard Deviation vs. Sepsis Patient Correlation Matrix (Cohort A)",
    cmap="bkr",
    clim=(-1, 1),
    width=2000,
    height=2000,
).opts(xrotation=90)

hv.save(heatmap, plot_outputs / "Sepsis Cohort A Standard Deviation Correlation Matrix.html")
```

```python
heatmap_stds
```

```python
std_correlation_with_sepsis = stds_correlation_matrix["sepsis_patient"] 
std_correlation_with_sepsis.columns = "sepsis_corr"

plt.figure(figsize=(21, 10))

std_correlation_with_sepsis.sort_values(ascending=False).plot(kind="bar")

plt.title("Sepsis Patient Correlation")
plt.xlabel("Features")
plt.ylabel("Correlation Coefficient")
plt.xticks(rotation=90, fontsize=8)
plt.tight_layout()
plt.show()
```

```python
bar_plot_stds = std_correlation_with_sepsis.sort_values(ascending=False).hvplot.bar(
    title="Standard Deviation vs. Sepsis Patient Correlation (Cohort A)",
    xlabel="Features",
    ylabel="Correlation Coefficient",
    rot=90,
    width=1800, 
    height=600
).opts(
    fontsize={
        "title": 20,
        "labels": 10,
        "xticks": 8,
        "yticks": 10
    }
)

hv.save(bar_plot_stds, plot_outputs / "Sepsis Cohort A Standard Deviation Correlation Bars.html")
```
