```python
import pandas as pd
from pathlib import Path
```

```python
INTER_DATA_PATH = "../intermediary_data/"
print(f"Loading and saving data from/to {Path(INTER_DATA_PATH).resolve()}")
```

```python
csv_filename = "vital_periodic_with_apache_all"
```

```python
vp_apache_path = Path(INTER_DATA_PATH) / Path(csv_filename + ".csv")

df_vp_apache_valid = pd.read_csv(vp_apache_path, engine="pyarrow")
```

```python
parquet_file_path = Path(INTER_DATA_PATH) / Path(csv_filename + ".parquet")

df_vp_apache_valid.to_parquet(path=parquet_file_path)
```
