```python
df_gtn_scalar = pd.read_csv(gtn_scalar_path)
df_sand_scalar = pd.read_csv(sand_scalar_path)
df_lstm_scalar = pd.read_csv(lstm_scalar_path)

```

```python
fig, axs, font_label, font_annots = figure_setup(num_row=1, num_cols=1, width_cm=14, height_cm=10)
# color_palette = palettable.matplotlib.Plasma_6.mpl_colors[:]
color_palette = palettable.tableau.PurpleGray_6.mpl_colors[:]

# Create the base bar plot without filling
bars = sns.barplot(x='Metric', y='Value', hue="model", data=melted_df, ax=axs, dodge=1, palette=color_palette)


# Add hatching manually using Matplotlib
hatches = ['/', "-", '.']  # Define different hatch patterns
n_bars = len(melted_df['Metric'].unique())
bar_width = 0.8 / len(melted_df["model"].unique())  # Calculate individual bar width

for i, patch in enumerate(axs.patches):
    # Apply hatching to each bar
    hatch = hatches[i // n_bars % len(hatches)]
    patch.set_hatch(hatch)
    patch.set_width(bar_width)


# Annotate bars with percentages and rotate labels by 90 degrees
for container in axs.containers:
    axs.bar_label(container, fontsize=8, fmt="{:.1f}%", label_type="edge", rotation=90, padding=2)

axs.set_ylim(0, 100)

# Set plot labels and title
axs.set_ylabel("Percentage")
axs.set_xlabel("Metrics")
axs.set_xticklabels(axs.get_xticklabels(), rotation=20, ha="right")
#axs.set_title("Performance Metrics Comparison Across Models with Hatching")

legend_hatches = ['////', "----", '....'] 
handles = []
for i, model in enumerate(combined_df["model"].unique()):
    handles.append(Patch(facecolor=color_palette[i], hatch=legend_hatches[i], label=model))
    
axs.legend(
    handles=handles,
    # ["Gated Transformer Network", "SAnD", "LSTM"],
    # title="Model",
    # facecolor="w",
    loc="upper center",
    bbox_to_anchor=(0.5, -0.4),
    ncol=3,
    frameon=False,
    fancybox=True,
    shadow=True,
    handleheight=2,
    handlelength=3,
    prop=font_annots,
)



plt.tight_layout()
plt.show()


if save_as_svg := True:
    fig.savefig(plot_output_dir / Path("model_performance_comp_scalar_hatching.svg"), format="svg")
```

```python
models_results = [df_gtn_prc, df_sand_prc, df_lstm_prc]  # Replace with your actual dataframes
model_names = ["GTN", "SAnD", "LSTM"]  # Names of your modelsA
# colors = ["#9345c9", "#2ca02c", "#d62728"] 
#colors = palettable.matplotlib.Plasma_6.mpl_colors[:]

colors = palettable.tableau.PurpleGray_6.mpl_colors[:]
```

```python
fig, axs_pr, font_label, font_annots = figure_setup(num_row=1, num_cols=1, width_cm=12, height_cm=8)

for i, (results, model_name, color) in enumerate(zip(models_results, model_names, colors)):

    pr_curve_x = results["x"][::-1]  # recall
    pr_curve_y = results["y"][::-1]  # precision
    
    auc = metrics.auc(pr_curve_x, pr_curve_y)

    sns.lineplot(
        x=pr_curve_x, y=pr_curve_y, ax=axs_pr, sort=False, estimator=None, color=color, label=f"{model_name} (AUC: {auc:.3f})", linewidth=2
    )

axs_pr.yaxis.set_major_locator(ticker.MultipleLocator(0.1))

axs_pr.set_ylabel("Precision")
axs_pr.set_xlabel("Recall")
axs_pr.set(xlim=(0, 1), ylim=(0, 1))

# axs_pr.legend(loc="best", fancybox=True, shadow=False, framealpha=0.2)

# axs_pr.set_title("Precision-Recall Curve Comparison Across Models")

axs_pr.legend(
    # handles=handles,
    # ["Gated Transformer Network", "SAnD", "LSTM"],
    # title="Model",
    # facecolor="w",
    loc="upper center",
    bbox_to_anchor=(0.5, -0.2),
    ncol=3,
    frameon=False,
    fancybox=True,
    shadow=True,
    #handleheight=1,
    #handlelength=3,
    prop=font_annots,
)

# Display the plot
plt.tight_layout()
plt.show()

if save_as_svg := True:
    fig.savefig(plot_output_dir / Path("model_prc_curve.svg"), format="svg")
```
