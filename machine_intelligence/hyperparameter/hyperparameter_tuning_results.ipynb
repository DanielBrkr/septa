```python
import pandas as pd
from pathlib import Path
from datetime import datetime
```

```python
cwd = Path.cwd()
cwd
```

```python
project_root = Path(cwd).parents[1]
project_root
```

```python
current_date = datetime.now().strftime("%Y-%m-%d")
plot_output_dir = project_root / "plot_outputs" / current_date
plot_output_dir.mkdir(parents=True, exist_ok=True)
```

```python
table_outputs = project_root / "table_outputs" / current_date
table_outputs.mkdir(parents=True, exist_ok=True)
```

```python
raw_gtn_results_path = cwd / Path("hyperparam_results_gtn.csv")
```

```python
df_raw_gtn_results = pd.read_csv(raw_gtn_results_path)
```

```python
df_raw_gtn_results_top_20 = df_raw_gtn_results.head(n=20)
df_raw_gtn_results_top_20
```

```python
gtn_parameters = ["params.batch_size_gpu", "model.net.bottleneck_factor", "model.net.d_model", "model.net.dropout",
                  "model.net.dropout_static", "model.net.num_classifier_head_hidden", "model.net.num_heads",
                  "model.net.num_layers", "model.net.num_static_hidden", "model.net.num_static_outs",
                  "optimizer.params.learning_rate", "optimizer.params.weight_decay", "train/auroc", "train/auprc",
                  "train/recall@80", "val/auroc", "val/auprc", "val/recall@80"]
```

```python
df_gtn_results_top_20 = df_raw_gtn_results_top_20.loc[:, gtn_parameters]
df_gtn_results_top_20
```

```python
df_gtn_results_top_20.columns = [
    "Batch Size", "Bottleneck Factor", "Model Dimension",
    "Dropout", "Static Dropout", "Classifier Head Hidden Units",
    "Attention Heads", "Layers", "Static Hidden Units",
    "Static Outputs", "Learning Rate", "Weight Decay",
    "Train AUROC", "Train AUPRC", "Train Precision@Recall80",
    "Validation AUROC", "Validation AUPRC", "Validation Precision@Recall80"
]
```

```python
df_gtn_results_top_20
```

```python
df_gtn_results_top_20.to_csv(table_outputs / "hyperparameter_tuning_results_gtn_top20.csv", index=False, sep=';',
                             decimal=".",
                             float_format='%.4f', )
```
