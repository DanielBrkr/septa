Very good explanation why we do that,
https://stackoverflow.com/questions/51030782/why-do-we-pack-the-sequences-in-pytorch

in a nutshell, packed padded sequences allow to batch sequences of varying length, while saving computations by transforming the individual sequences to one 
long sequence that is interleaved. So just a different representation, that allows pytorch to be more efficient, even to a very large extent (dropping computation times),
considering the "blank" compute just to make the sequences the same length.


```python
#a = torch.ones(10)
a = torch.linspace(10,14, steps=5)
#a_labels = torch.ones(10)
a_labels = torch.linspace(10,14, steps=5)
print(a)
print(a_labels)
```

```python
len(a)
```

```python
b = torch.linspace(1, 6, steps=6)
b_labels = torch.linspace(1, 6, steps=6)
print(b)
print(b_labels)
```

```python
c = torch.linspace(101, 107, steps=7)
c_labels = torch.linspace(101, 107, steps=7)

```

```python
a_b_c_padded = torch.nn.utils.rnn.pad_sequence([a, b, c], batch_first=True)
a_b_c_padded
```

```python
a_b_packed_padded = torch.nn.utils.rnn.pack_padded_sequence(a_b_c_padded, lengths=[len(a), len(b), len(c)], batch_first=True, enforce_sorted=False)
a_b_packed_padded
```

```python
a_b_c_labels_padded = torch.nn.utils.rnn.pad_sequence([a_labels, b_labels, c_labels], batch_first=True)
a_b_c_labels_packed = torch.nn.utils.rnn.pack_padded_sequence(a_b_c_labels_padded, lengths=[len(a), len(b), len(c)], batch_first=True, enforce_sorted=False)

print(a_b_c_labels_padded)
print(a_b_c_labels_packed)
```

Now applying the inverse operation of pack_padded_sequence

```python
a_b_c_unpacked = torch.nn.utils.rnn.pad_packed_sequence(a_b_packed_padded, batch_first=True)
a_b_c_labels_unpacked = torch.nn.utils.rnn.pad_packed_sequence(a_b_c_labels_packed, batch_first=True)
print(a_b_c_unpacked)
print(a_b_c_labels_unpacked)
```
