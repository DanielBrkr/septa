```python
import shap
import xgboost

```

```python

# train XGBoost model
X,y = shap.datasets.adult()
model = xgboost.XGBClassifier(nestimators=100, max_depth=2).fit(X, y)
```

```python
X.shape
```

```python
# compute SHAP values
explainer = shap.Explainer(model, X)
shap_values = explainer(X[:1000])
```
