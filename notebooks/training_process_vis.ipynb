```python
gtn_rep_path = Path("gtn_training_report.csv")
sand_rep_path = Path("sand_training_report.csv")
lstm_rep_path = Path("lstm_training_report.csv")

```

```python
df_gtn = pd.read_csv(gtn_rep_path)
df_sand = pd.read_csv(sand_rep_path)
df_lstm = pd.read_csv(lstm_rep_path)
df_gtn.head()
```

```python
metric_columns = ["train/acc", "val/acc", "train/recall", "val/recall", "train/spec", "val/spec", "train/prec",
                  "val/prec", "train/recall@80", "val/recall@80", "train/auroc", "val/auroc", "train/auprc",
                  "val/auprc", "train/loss_epoch", "val/loss_epoch"]
```

```python
run_name_gtn = "eternal-sponge-8 - "
run_name_sand = "misunderstood-jazz-12 - "
run_name_lstm = "giddy-darkness-4 - "

```

```python
run_metric_columns_gtn = [run_name_gtn + metric for metric in metric_columns]
run_metric_columns_sand = [run_name_sand + metric for metric in metric_columns]
run_metric_columns_lstm = [run_name_lstm + metric for metric in metric_columns]

```

```python
df_gtn_metrics = df_gtn[["epoch", *run_metric_columns_gtn]]
df_sand_metrics = df_sand[["epoch", *run_metric_columns_sand]]
df_lstm_metrics = df_lstm[["epoch", *run_metric_columns_lstm]]

```

```python
df_gtn_metrics.columns = ["epoch", *metric_columns]
df_sand_metrics.columns = ["epoch", *metric_columns]
df_lstm_metrics.columns = ["epoch", *metric_columns]

```

```python
df_gtn_metrics.plot(x="epoch")
```

```python
title_list = ["Accuracy", "Recall/Sensitivity", "Specificity", "Precision", "Precision@Recall80","AUROC", "AUPRC",
              "Loss"]
```

```python
import palettable

fig, axs, font_labels, font_annots = figure_setup(num_row=4, num_cols=2, height_cm=16, share_x=True,
                              palette=palettable.tableau.PurpleGray_6.mpl_colors[:])

axs_flt = axs.flatten()
# df_gtn_metrics[["epoch", "train/acc", "val/acc"]].plot(ax=axs_gtn[0,0])
# df_gtn_metrics.plot(x="epoch", y=metric_columns[0:2], ax=axs_ftl[])

# for idx, _ in enumerate(metric_columns[::2]):
for idx, _ in enumerate(metric_columns[::2]):
    # if idx % 2 == 0:
    df_gtn_metrics.plot(x="epoch", y=metric_columns[idx * 2:idx * 2 + 2], ax=axs_flt[idx], title=title_list[idx])


for ax in axs_flt:
    # ax.set_ylim(0,1)
    ax.get_legend().set_visible(False)

plt.tight_layout()


axs_flt[-1].legend(
    ["Training", "Validation"],
    loc="upper center",
    bbox_to_anchor=(-0.1, -0.4),
    ncol=2,
    frameon=False,
    fancybox=True,
    shadow=True,
    prop=font_annots,
)


if save_as_svg := True:
    fig.savefig(plot_output_dir / Path("gtn_training_process.svg"), format="svg")
```
